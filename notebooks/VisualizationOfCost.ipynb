```kotlin
import hu.akos.hollo.szabo.math.Permutation
import hu.akos.hollo.szabo.logic.logging.LogLoader
import hu.akos.hollo.szabo.logic.logging.loadByteFile
import hu.akos.hollo.szabo.model.logging.LogEvent
import hu.akos.hollo.szabo.model.solution.SolutionWithIterationAndIndex
import kotlin.io.path.Path
import kotlin.reflect.typeOf

val loader = LogLoader(::loadByteFile)
val basePath = Path("D:\\Research\\results\\2025-02-01T22_10_53_436476800")

basePath.toFile().listFiles().asSequence()
    .filter { it.name.matches("[0-9]+-solution".toRegex()) }
    .mapNotNull { file ->
        val result = loader.load<SolutionWithIterationAndIndex<Permutation>>(
            Path(file.path),
            typeOf<LogEvent<SolutionWithIterationAndIndex<Permutation>>>()
        )
        result.message.cost?.get(0)
    }
    .let {
        var best = Float.MAX_VALUE
        it
            .filter { it < best }
            .forEach {
                best = it
                println(it)
            }
    }
```
