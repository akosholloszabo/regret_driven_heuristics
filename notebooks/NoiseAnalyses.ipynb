```kotlin
import hu.akos.hollo.szabo.model.loadRecords
val records = loadRecords("../output/output-2025-09-21.txt")
records.last()
```

```kotlin
import org.jetbrains.kotlinx.dataframe.api.dataFrameOf
import org.jetbrains.kotlinx.dataframe.api.toDataFrame
val recordsFiltered = records
    .filter {
        it.threeOptIteration <= 5 && it.threeOptIteration >= 0
    }
val dataframe = dataFrameOf(
    "noiseLevel" to recordsFiltered.map { it.noiseLevel * 100 },
    "gapAverages" to recordsFiltered.map { it.gapRatioStats.avg * 100 },
    "threeOptIterations" to recordsFiltered.map { it.threeOptIteration },
    "min" to recordsFiltered.map { 0 },
)
```

```kotlin
import org.jetbrains.kotlinx.dataframe.api.filter
import org.jetbrains.letsPlot.ggsize

plotGrid(listOf(
    dataframe.plot {
        groupBy("threeOptIterations"){
            ribbon {
                x("noiseLevel") { axis.name = "Noise Level (%)" }
                yMin("min")
                yMax("gapAverages")
                fillColor("threeOptIterations") {
                    legend.type = LegendType.None
                    // reuse same categorical palette as the line colors
                    scale = categorical(
                        listOf(
                            Color.hex("#0044dd"),
                            Color.hex("#3344aa"),
                            Color.hex("#664499"),
                            Color.hex("#994466"),
                            Color.hex("#aa4433"),
                            Color.hex("#dd4400"),
                        ),
                        listOf(0, 5)
                    )
                    alpha = 0.18
                }
            }
            line {
                x("noiseLevel") { axis.name = "Noise Level (%)" }
                y("gapAverages") {
                    axis.name = "Gap Ratio Average (%)"
                }
                color("threeOptIterations") {
                    legend.type = LegendType.None
                    scale = categorical(
                        listOf(
                            Color.hex("#0044dd"),
                            Color.hex("#3344aa"),
                            Color.hex("#664499"),
                            Color.hex("#994466"),
                            Color.hex("#aa4433"),
                            Color.hex("#dd4400"),
                        ),
                        listOf(0, 5)
                    )
                }
            }
        }
        layout.title = "Gap Ratio Average per Noise Level"// on Single Solution Instances"
    },
    dataframe
        .filter { (it["threeOptIterations"] as Int) > 0 }
        .plot {
        groupBy("threeOptIterations"){
            ribbon {
                x("noiseLevel") { axis.name = "Noise Level (%)" }
                yMin("min")
                yMax("gapAverages")
                fillColor("threeOptIterations") {
                    // reuse same categorical palette as the line colors
                    scale = categorical(
                        listOf(
                            Color.hex("#0044dd"),
                            Color.hex("#3344aa"),
                            Color.hex("#664499"),
                            Color.hex("#994466"),
                            Color.hex("#aa4433"),
                            Color.hex("#dd4400"),
                        ),
                        listOf(0, 5)
                    )
                    alpha = 0.18
                    legend.type = LegendType.None
                }
            }
            line {
                x("noiseLevel") { axis.name = "Noise Level (%)" }
                y("gapAverages") {
                    axis.name = "Gap Ratio Average (%)"
                }
                color("threeOptIterations") {
                    legend.name = "RG-PE-3Opt Iterations"
                    legend.breaksLabeled(
                        0 to "none",
                        1 to "1 iteration",
                        2 to "2 iterations",
                        3 to "3 iterations",
                        4 to "4 iterations",
                        5 to "5 iterations",
                    )
                    scale = categorical(
                        listOf(
                            Color.hex("#0044dd"),
                            Color.hex("#3344aa"),
                            Color.hex("#664499"),
                            Color.hex("#994466"),
                            Color.hex("#aa4433"),
                            Color.hex("#dd4400"),
                        ),
                        listOf(0, 5)
                    )
                }
            }
        }
        layout.title = "Gap Ratio Average per Noise Level"// on Single Solution Instances"
    }),
    nCol = 2,
    widths = listOf(2,3)
)
```
